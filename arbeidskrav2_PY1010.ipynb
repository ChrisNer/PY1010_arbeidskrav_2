<h3>Oppg 2)</h3> 
Det skal arrangeres en klassefest og man antar at hver elev spiser 1/4 pizza. Lag et
program som tar inn antall elever fra konsollen ved
antall_elever = int(input('Skriv inn antall elever:' ))

Programmet skal så regne ut hvor mange pizzaer som skal handles inn til festen og skrive
svaret til skjerm. Merk, man kan ikke kjøpe 4 og en kvart pizza på butikken (man må da kjøpe
5).

Hint1: Gir programmet ditt et fornuftig svar hvis det f.eks er 21 elever i klassen?
Hint2: Det er ikke vanlig å si/skrive: ‘Det må handles inn 6.0 pizzaer til festen’. Hvordan kan
sikre at antall pizzaer skrives ut som et heltall (ikke desimaltall)?

```python
# Importere math
import math

# Angi antall elever.
antall_elever = int(input('Skriv inn antall elever:' ))

# Hver elev spiser 1/4 pizza.
pizza_per_elev = 0.25

# Bergene antall pizzaer, rundet opp.
antall_pizza = math.ceil(antall_elever * pizza_per_elev)

#Printe resultat.
print(f"""{antall_elever} kommer elever til klassefesten. 
De må kjøpe {antall_pizza} pizzaer, slik alle får like mye.""")
```

<h3>Oppg 3)</h3>
Lag et program med en funksjon som regner om fra grader til radianer.
Programmet skal starte med:
import numpy as np
v_grad = float(input('Skriv inn gradtallet:' ))
Radiantallet til vinkelen regnes så ut ved følgende formel: v_rad = v_grad*np.pi/180
Resultatet v_rad skrives til skjerm med passende tekst og verdi.
Merk: np.pi er en ferdiglaget funksjon som gir verdien 3.1415....

```python
# Import av numpy
import numpy as np

# Funksjon for å beregne radiantallet.
def beregn_radian(vinkel_grader):
    v_rad = vinkel_grader * np.pi/180
    return v_rad

# Angi antall grader.
v_grad = float(input('Skriv inn gradtallet:' ))

# Printa resultat.
print(f"Angitt vinkel er {v_grad} grader, radiantallet er {beregn_radian(v_grad)}")
```

c) Lag et program som ber brukeren skrive inn info om et nytt land (altså et land som
ikke allerede finnes i dictionaryen data). Videre skal brukeren oppgi hovedstad og
antall innbyggere for det «nye» landet. Programmet skal så utvide/oppdatere

```python
# Del A) Dictionary med angitt data.
data = {
    "Norge": ["Oslo", 0.634],
    "England": ["London", 8.982],
    "Frankrike": ["Paris", 2.161],
    "Italia": ["Roma", 2.873]
    }
```

```python
# Del B) 
# Spør om land som existerer på listen.
while True:  
    # User input. Angi et land.
    bruker_land = input("Skriv nanvet på et land.").title()

    # Sjekk hvis landet finnes i data.
    if bruker_land in data:
        break   # Bryt loop. Landet finnes i data.
    else:
        print(f"{bruker_land} finnes ikke på listen. Prøv igjen med et annet land.")

# Hente data for valgt land.
hovedstad, inbyggere = data[bruker_land]

# Presentere resultat.
print(f"""{hovedstad} er hovedstaden i {bruker_land} 
og det er {inbyggere} mill. innbyggere i {hovedstad}.""")
```

```python
# Del C)
# Brukeren angir ny data.
nytt_land = input("Angi et nytt land: ").title()
ny_hovedstad = input(f"Angi navnet på hovedstaden i {nytt_land}: ").title()
antall_innbyggere = float(input(f"Angi antall innbyggere i {ny_hovedstad} i antall millioner. (F. eks. 3.214): "))

# Oppdatere data med nytt land.
data[nytt_land] = [ny_hovedstad, round(antall_innbyggere, 3)] 

# Presentere info om nytt land.
print(f"Info i data er oppdatert med nytt land: {nytt_land}")

# Presentera data
print(data)
```

<h3>Oppg 5)</h3> (BILDE I DOKUMENT MED OPPGAVER!) 
Lag et program med en funksjon som tar a og b som inn-argumenter og som så
regner ut arealet og «ytre» omkrets til en figur satt sammen av en rettvinklet trekant og en
halvsirkel, se figuren under. Med «ytre» omkrets menes samlet lengde av de sorte strekene.
Funksjonen skal returnere arealet og «ytre» omkrets, som så skrives til skjerm med passende
tekst.

```python
import numpy as np

# Funksjon som beregner arealet og omkrekts til figuren.
def beregn_figur(a, b):
    # Beregn areal.
    figur_areal = ((np.pi * a**2) / 8) + ((a * b) / 2)

    # Beregn figurens omkrets.
    figur_omkrets = b + (np.sqrt(a**2 + b**2)) + ((np.pi * a) / 2)

    return figur_areal, figur_omkrets

# Hente info om a og b.
a_lengde = float(input("Skriv lengde på side a: "))
b_lengde = float(input("Skriv lengde på side b: "))

# Bruk ca
areal, omkrets = beregn_figur(a_lengde, b_lengde)

# Presentere resultater.
print(f"""Du har angitt lengden {a_lengde} på side a og lengden {b_lengde} på side b.
Dette gir figuren et areal på {areal:.2f} og en omkrets på {omkrets:.2f}.""")
```

<h3>Oppg 6)</h3> 
Skriv en kode som plotter funksjonen 𝑓(𝑥) = −𝑥2 − 5, for x på intervallet [-10,10].
Hint: np.linspace(-10, 10, 200) gir en array med 200 punkter jevnt fordelt på intervallet
[-10,10].

```python
# Import av pakker.
import numpy as np
import matplotlib.pyplot as plt

# Definiere x-intervallet.
x = np.linspace(-10, 10, 200)

# Definiera funktionen f(x)
y = -(x**2) - 5

# Plot.
plt.plot(x, y, label=r"$f(x) = -x^2 - 5$", color="blue")

# Styling.
plt.title("Grafen til funksjonen $f(x) = -x^2 - 5$")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()

plt.show()
```
